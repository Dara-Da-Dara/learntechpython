# DSPy + Gradio Deployment Notebook

This notebook shows how to build a DSPy program and deploy it as a Gradio web app.

## 1. Install Dependencies

## 2. Configure DSPy LLM

```
import dspy

# Configure your LLM (OpenAI-compatible)
llm = dspy.OpenAI(
    model="gpt-4o-mini",
    api_key="YOUR_API_KEY"
)

dspy.settings.configure(llm=llm)
```

## 3. Define DSPy Signature

```
class QuestionAnswer(dspy.Signature):
    """Answer questions clearly and concisely."""
    question = dspy.InputField(desc="User question")
    answer = dspy.OutputField(desc="Helpful answer")
```

## 4. Define DSPy Module

```
class QAProgram(dspy.Module):
    def __init__(self):
        super().__init__()
        self.predict = dspy.Predict(QuestionAnswer)

    def forward(self, question):
        return self.predict(question=question)

qa_program = QAProgram()
```

## 5. Test DSPy Program

```
result = qa_program("What is agentic AI?")
print(result.answer)
```

## 6. Wrap DSPy Program with Gradio

```
import gradio as gr

def chat_fn(user_input):
    response = qa_program(user_input)
    return response.answer
```

## 7. Launch Gradio App

```
demo = gr.Interface(
    fn=chat_fn,
    inputs=gr.Textbox(lines=2, placeholder="Ask a question..."),
    outputs=gr.Textbox(label="Answer"),
    title="DSPy + Gradio QA App",
    description="A simple question-answering app powered by DSPy"
)

demo.launch()
```
