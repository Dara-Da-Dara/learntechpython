This is not a chatbot — it’s a single-turn Q&A system.

```python
User Question
     ↓
Gradio Web UI
     ↓
Pretrained Q&A Model (FLAN-T5)
     ↓
Generated Answer
     ↓
Displayed in Gradio
```

| Component                     | Purpose                         |
| ----------------------------- | ------------------------------- |
| **Google Colab**              | Free cloud environment with GPU |
| **Gradio**                    | Web interface for Q&A           |
| **Hugging Face Transformers** | Load pretrained language model  |
| **FLAN-T5**                   | Instruction-tuned Q&A model     |
| **PyTorch**                   | Model inference                 |


| **Step No.** | **Code Section**                          | **What Happens**                  | **Why It Is Needed**                     |
| ------------ | ----------------------------------------- | --------------------------------- | ---------------------------------------- |
| 1            | `import gradio as gr`                     | Imports Gradio library            | To create the web UI                     |
| 2            | `import torch`                            | Imports PyTorch                   | Required to run deep learning models     |
| 3            | `from transformers import ...`            | Imports tokenizer & model classes | To load pretrained Q&A model             |
| 4            | `MODEL_NAME = "google/flan-t5-base"`      | Selects the pretrained model      | Defines which Q&A model to use           |
| 5            | `AutoTokenizer.from_pretrained()`         | Loads tokenizer                   | Converts text → tokens & tokens → text   |
| 6            | `AutoModelForSeq2SeqLM.from_pretrained()` | Loads language model              | Generates answers from questions         |
| 7            | `device_map="auto"`                       | Auto-selects CPU/GPU              | Improves performance                     |
| 8            | `torch_dtype=torch.float16`               | Uses half-precision               | Saves memory & speeds up inference       |
| 9            | `def qa_model(question):`                 | Defines Q&A function              | Core logic that Gradio calls             |
| 10           | `prompt = f"Answer the question..."`      | Creates instruction prompt        | Helps model understand the task          |
| 11           | `tokenizer(prompt, ...)`                  | Tokenizes the input question      | Converts text into model-readable format |
| 12           | `.to(model.device)`                       | Moves data to CPU/GPU             | Ensures model & data are on same device  |
| 13           | `model.generate()`                        | Generates the answer              | Produces output tokens                   |
| 14           | `max_new_tokens=128`                      | Limits answer length              | Prevents overly long outputs             |
| 15           | `temperature=0.3`                         | Controls randomness               | Lower value → more factual answers       |
| 16           | `tokenizer.decode()`                      | Converts tokens to text           | Makes answer readable                    |
| 17           | `return answer`                           | Sends answer back                 | Required for Gradio output               |
| 18           | `gr.Interface()`                          | Creates UI layout                 | Connects model to web interface          |
| 19           | `inputs=gr.Textbox()`                     | Input component                   | Allows user to type question             |
| 20           | `outputs=gr.Textbox()`                    | Output component                  | Displays model answer                    |
| 21           | `title & description`                     | Adds UI text                      | Improves usability                       |
| 22           | `.launch(share=True)`                     | Starts the app                    | Generates public Gradio link             |


```python
import gradio as gr
import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

MODEL_NAME = "google/flan-t5-base"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForSeq2SeqLM.from_pretrained(
    MODEL_NAME,
    device_map="auto",
    torch_dtype=torch.float16
)

def qa_model(question):
    prompt = f"Answer the question:\n{question}"

    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

    outputs = model.generate(
        **inputs,
        max_new_tokens=128,
        temperature=0.3
    )

    answer = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )
    return answer

gr.Interface(
    fn=qa_model,
    inputs=gr.Textbox(lines=2, placeholder="Enter your question"),
    outputs=gr.Textbox(label="Answer"),
    title="❓ Q&A Model (Gradio + Colab)",
    description="Ask a question and get a direct answer"
).launch(share=True)
```
