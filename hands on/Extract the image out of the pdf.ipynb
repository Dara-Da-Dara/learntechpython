```python
import fitz  # PyMuPDF
import cv2
import os
import numpy as np

def apply_colormap(img_gray):
    # Apply colormap to the grayscale image
    img_color = cv2.applyColorMap(img_gray, cv2.COLORMAP_VIRIDIS)
    return img_color

def extract_images_and_edges(document_path, output_folder):
    # Create output folder if it doesn't exist
    os.makedirs(output_folder, exist_ok=True)

    # Open the document
    pdf_document = fitz.open(document_path)

    for page_num in range(pdf_document.page_count):
        page = pdf_document[page_num]

        # Get images on the page
        images = page.get_images(full=True)

        for img_index, img_info in enumerate(images):
            image_index = img_info[0]
            base_image = pdf_document.extract_image(image_index)
            image_bytes = base_image["image"]

            # Convert image bytes to a numpy array
            img_gray = cv2.imdecode(
                np.frombuffer(image_bytes, np.uint8), cv2.IMREAD_GRAYSCALE
            )

            # Apply colormap to the grayscale image
            img_color = apply_colormap(img_gray)

            # Save the colored image
            color_image_filename = f"{output_folder}/page{page_num + 1}_img{img_index + 1}_color.png"
            cv2.imwrite(color_image_filename, img_color)

            # Apply Canny edge detection
            edges = cv2.Canny(img_gray, 100, 200)

            # Save the edge image
            edges_filename = f"{output_folder}/page{page_num + 1}_img{img_index + 1}_edges.png"
            cv2.imwrite(edges_filename, edges)

    # Close the PDF document
    pdf_document.close()

# Example usage with corrected file paths
document_path = "C:/Users/hp/Documents/year wise data/2024/llm_generative_ai/pdf_to_image_extration/ahsr101.pdf"
output_folder_path = "C:/Users/hp/Documents/year wise data/2024/llm_generative_ai/pdf_to_image_extration/new_code"
extract_images_and_edges(document_path, output_folder_path)
```
